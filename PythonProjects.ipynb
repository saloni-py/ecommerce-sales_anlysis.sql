Password Generator

```python
def validate_password(password):
  errors:list[str]=[]
  if len(password)<8:
    print("minimum 8 characters required")
  if not any(ch.isupper() for ch in password):
    print("atleast one uppercase letter required")
  if not any(ch.islower() for ch in password):
    print("atleast one lowercase letter required")
  if not any(ch.isdigit() for ch in password):
    print("atleast one digit required")
  special_character="!@#$%^&*"
  if not any(ch in special_character for ch in password):
    print("atleast one special character required")

  if not errors:
    return True, []
  else:
    return False, errors

password=input("enter your password:")
valid,unmet=validate_password(password)
if valid:
  print("password is valid")
else:
  print("password is invalid")
  print("unmet requirements:")
  for error in unmet:
    print(error)
```

Employee Payroll Management System

```python
class Employee:
    def __init__(self, emp_id, name, basic_salary, allowance, deduction):
        self.emp_id = emp_id
        self.name = name
        self.basic_salary = basic_salary
        self.allowance = allowance
        self.deduction = deduction

    def gross_salary(self):
        return self.basic_salary + self.allowance

    def net_salary(self):
        return self.gross_salary() - self.deduction

    def update_salary(self, new_salary):
        self.basic_salary = new_salary


employees = []


def add_employee():
    emp_id = input("Enter Employee ID: ")

    for employee in employees:
        if employee.emp_id == emp_id:
            print("Employee ID already exists!")
            return

    name = input("Enter Employee Name: ")
    basic_salary = float(input("Enter Basic Salary: "))
    allowance = float(input("Enter Allowance: "))
    deduction = float(input("Enter Deduction: "))

    employee = Employee(
        emp_id,
        name,
        basic_salary,
        allowance,
        deduction
    )

    employees.append(employee)

    print("Employee added successfully!")


def display_employees():
    if len(employees) == 0:
        print("No employees found!")
        return

    for employee in employees:
        print("\n-------------------------")
        print("ID:", employee.emp_id)
        print("Name:", employee.name)
        print("Basic Salary:", employee.basic_salary)
        print("Allowance:", employee.allowance)
        print("Deduction:", employee.deduction)
        print("Gross Salary:", employee.gross_salary())
        print("Net Salary:", employee.net_salary())


def search_employee():
    emp_id = input("Enter Employee ID to search: ")

    for employee in employees:
        if employee.emp_id == emp_id:
            print("\nEmployee Found!")
            print("ID:", employee.emp_id)
            print("Name:", employee.name)
            print("Basic Salary:", employee.basic_salary)
            print("Allowance:", employee.allowance)
            print("Deduction:", employee.deduction)
            print("Gross Salary:", employee.gross_salary())
            print("Net Salary:", employee.net_salary())
            return

    print("Employee not found!")


def update_salary():
    emp_id = input("Enter Employee ID: ")

    for employee in employees:
        if employee.emp_id == emp_id:
            new_salary = float(input("Enter New Basic Salary: "))

            employee.update_salary(new_salary)

            print("Salary updated successfully!")
            return

    print("Employee not found!")


def payroll_summary():
    if len(employees) == 0:
        print("No employees found!")
        return

    total_gross = 0
    total_net = 0

    print("\n========== PAYROLL SUMMARY ==========")

    for employee in employees:
        gross = employee.gross_salary()
        net = employee.net_salary()

        print(
            employee.emp_id,
            employee.name,
            "Gross:", gross,
            "Net:", net
        )

        total_gross += gross
        total_net += net

    print("-------------------------------------")
    print("Total Gross Salary:", total_gross)
    print("Total Net Salary:", total_net)

while True:

    print("\n========== PAYROLL SYSTEM ==========")
    print("1. Add Employee")
    print("2. Display Employees")
    print("3. Search Employee")
    print("4. Update Salary")
    print("5. Payroll Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_employee()

    elif choice == "2":
        display_employees()

    elif choice == "3":
        search_employee()

    elif choice == "4":
        update_salary()

    elif choice == "5":
        payroll_summary()

    elif choice == "6":
        print("Thank you for using Payroll System!")
        break

    else:
        print("Invalid choice! Please try again.")
```
